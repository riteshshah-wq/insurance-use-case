# 01 - Data ExplorationExplore sample insurance claims and policy documents used in this demo.

```python
import jsonfrom pathlib import Path# Load sample claimsclaims_dir = Path('../data/sample_claims')for f in sorted(claims_dir.glob('*.txt')):    text = f.read_text()    print(f'=== {f.name} ({len(text)} chars) ===')    print(text[:200] + '...')    print()
```

## Policy DocumentsThese are used by the RAG-powered customer agent.

```python
policies_dir = Path('../data/sample_policies')for f in sorted(policies_dir.glob('*.txt')):    text = f.read_text()    print(f'=== {f.name} ===')    print(f'Length: {len(text)} chars, {len(text.split())} words')    # Show sections    for line in text.split('\n'):        if line.startswith('SECTION') or line.startswith('COVERAGE'):            print(f'  {line.strip()}')    print()
```

## Applicant ProfilesTest data for the underwriting use case.

```python
profiles_dir = Path('../data/applicant_profiles')for f in sorted(profiles_dir.glob('*.json')):    profile = json.loads(f.read_text())    print(f"{profile['name']}: age={profile['age']}, type={profile['coverage_type']}, amount=${profile['coverage_amount']:,}")    if profile['health_conditions']:        print(f"  Health: {', '.join(profile['health_conditions'])}")    if profile.get('driving_record'):        print(f"  Driving: {profile['driving_record']}")    print()
```
