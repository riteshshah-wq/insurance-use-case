```python
# Simple retrieval metricsdef evaluate_retrieval(query, expected_doc):    results = retrieve(query, top_k=3)    docs_found = [r['document'] for r in results]    hit = expected_doc in docs_found    rank = docs_found.index(expected_doc) + 1 if hit else -1    return hit, ranktest_cases = [    ('What is my auto deductible?', 'auto_policy.txt'),    ('home insurance claim process', 'home_policy.txt'),    ('life insurance beneficiary', 'life_policy.txt'),]hits = 0for query, expected in test_cases:    hit, rank = evaluate_retrieval(query, expected)    hits += int(hit)    status = f'✓ rank={rank}' if hit else '✗ not found'    print(f'{status} | {query} → expected: {expected}')print(f'\nHit rate: {hits}/{len(test_cases)} ({hits/len(test_cases):.0%})')
```
